```python

LAST_NAME = "Ganade"
SEED_NUM = 4
FAVORITE_ARTIST = "Grentperez"

def char_position(character):
    return ord(character.upper()) - ord("A") + 1

artist_positions = [char_position(ch) for ch in FAVORITE_ARTIST if ch.isalpha()]
if not artist_positions:
    raise ValueError("FAVORITE_ARTIST must contain at least one letter.")

print("Student-specific inputs")
print(f"LAST_NAME: {LAST_NAME}")
print(f"SEED_NUM: {SEED_NUM}")
print(f"FAVORITE_ARTIST: {FAVORITE_ARTIST}")
print(f"Artist character positions: {artist_positions}")
```

```python
#[EXERCISE 1]

from collections import Counter, defaultdict
import pandas as pd

test_sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]
measurement_categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

test_records = []
for source_idx, source in enumerate(test_sources):
    for category_idx, category in enumerate(measurement_categories):
        artist_value = artist_positions[(source_idx + category_idx) % len(artist_positions)]
        value = SEED_NUM + artist_value + (source_idx + 1) * (category_idx + 1)
        test_records.append({
            "source": source,
            "category": category,
            "value": value
        })

test_records.append(test_records[0].copy())

generated_df = pd.DataFrame(test_records)
print("GENERATED TEST DATA")
print(generated_df.to_string(index=False))

organized_df = generated_df.sort_values(["source", "category", "value"]).reset_index(drop=True)
print("\nORGANIZED TEST DATA")
print(organized_df.to_string(index=False))

pair_counts = Counter(zip(generated_df["source"], generated_df["category"]))
repeated_pairs = {pair: count for pair, count in pair_counts.items() if count > 1}
distinct_pairs = sorted(pair_counts.keys())
print("\nREPEATED SOURCE/CATEGORY RECORDS")
print(repeated_pairs)
print("\nDISTINCT SOURCE/CATEGORY PAIRS")
print(distinct_pairs)

print("\nNUMERICAL SUMMARY")
print(generated_df["value"].describe().to_string())
print("\nCATEGORY SUMMARY (record counts)")
print(generated_df["category"].value_counts().sort_index().to_string())
print("\nMEAN VALUE BY CATEGORY")
print(generated_df.groupby("category")["value"].mean().sort_index().to_string())

processed_df = generated_df.groupby(["source", "category"], as_index=False).agg(
    measurement_count=("value", "size"),
    total_value=("value", "sum"),
    mean_value=("value", "mean")
)
ordered_df = processed_df.sort_values(["category", "source"]).reset_index(drop=True)
print("\nPROCESSED RESULTS")
print(ordered_df.to_string(index=False))

print("\nFINAL SUMMARY")
print(f"Total records: {len(generated_df)}")
print(f"Distinct sources: {generated_df['source'].nunique()}")
print(f"Distinct categories: {generated_df['category'].nunique()}")
print(f"Distinct source/category pairs: {len(distinct_pairs)}")
print(f"Repeated pairs: {repeated_pairs}")
print(f"Overall mean measurement value: {generated_df['value'].mean():.2f}")

print("\nFINAL ORGANIZED AND SUMMARIZED TEST DATASET")
print(ordered_df.to_string(index=False))
```

```python
# [EXERCISE 2]

student_names = [f"{LAST_NAME}_S{i}" for i in range(1, 6)]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

performance_records = []
for student_idx, student in enumerate(student_names):
    for subject_idx, subject in enumerate(subjects):
        artist_value = artist_positions[(student_idx + subject_idx) % len(artist_positions)]
        score = 50 + ((SEED_NUM * 7 + artist_value * 3 + student_idx * 5 + subject_idx * 4) % 51)
        performance_records.append({
            "student": student,
            "subject": subject,
            "score": score
        })


performance_records.append(performance_records[0].copy())


performance_records.append({
    "student": student_names[-1],
    "subject": "ENGLISH",
    "score": None
})

performance_df = pd.DataFrame(performance_records)
print("GENERATED PERFORMANCE RECORDS")
print(performance_df.to_string(index=False))


organized_performance = performance_df.sort_values(
    ["student", "subject"], na_position="last"
).reset_index(drop=True)
print("\nORGANIZED STUDENT/SUBJECT DATA")
print(organized_performance.to_string(index=False))

valid_pairs = performance_df.dropna(subset=["score"])
pair_counts = Counter(zip(valid_pairs["student"], valid_pairs["subject"]))
repeated_performance = {pair: count for pair, count in pair_counts.items() if count > 1}
incomplete_records = performance_df[performance_df["score"].isna()]
print("\nREPEATED STUDENT/SUBJECT RECORDS")
print(repeated_performance)
print("\nINCOMPLETE RECORDS")
print(incomplete_records.to_string(index=False))


def performance_category(score):
    if score >= 90:
        return "EXCELLENT"
    if score >= 80:
        return "VERY GOOD"
    if score >= 70:
        return "GOOD"
    if score >= 60:
        return "SATISFACTORY"
    return "NEEDS IMPROVEMENT"

performance_df["category"] = performance_df["score"].apply(
    lambda score: performance_category(score) if pd.notna(score) else "INCOMPLETE"
)
print("\nPERFORMANCE CATEGORIES")
print(performance_df.to_string(index=False))

valid_scores = performance_df.dropna(subset=["score"]).copy()
student_summary = valid_scores.groupby("student")["score"].agg(
    assessment_count="count", average_score="mean", highest_score="max", lowest_score="min"
).reset_index().sort_values("student")
subject_summary = valid_scores.groupby("subject")["score"].agg(
    assessment_count="count", average_score="mean", highest_score="max", lowest_score="min"
).reset_index().sort_values("subject")

print("\nSTUDENT SUMMARIES")
print(student_summary.to_string(index=False, formatters={"average_score": "{:.2f}".format}))
print("\nSUBJECT SUMMARIES")
print(subject_summary.to_string(index=False, formatters={"average_score": "{:.2f}".format}))

ordered_performance = performance_df.sort_values(
    ["student", "subject"], na_position="last"
).reset_index(drop=True)
print("\nORDERED RESULTS")
print(ordered_performance.to_string(index=False))

print("\nFINAL STUDENT PERFORMANCE REPORT")
print(f"Total records: {len(performance_df)}")
print(f"Valid scores: {len(valid_scores)}")
print(f"Incomplete records: {len(incomplete_records)}")
print(f"Repeated student/subject pairs: {repeated_performance}")
print(f"Overall average of valid scores: {valid_scores['score'].mean():.2f}")
print("\nStudent averages:")
print(student_summary[["student", "average_score"]].to_string(index=False, formatters={"average_score": "{:.2f}".format}))
print("\nSubject averages:")
print(subject_summary[["subject", "average_score"]].to_string(index=False, formatters={"average_score": "{:.2f}".format}))
```

```python
#[EXERCISE 3]
customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]
product_prices = {
    "NOTEBOOK": 35.00,
    "PEN": 12.00,
    "USB": 350.00,
    "HEADPHONES": 550.00,
    "CALCULATOR": 275.00
}
products = list(product_prices.keys())

transaction_records = []

for i in range(10):
    customer = customers[i % len(customers)]
    product_idx = (i + SEED_NUM + artist_positions[i % len(artist_positions)]) % len(products)
    product = products[product_idx]
    quantity = 1 + ((SEED_NUM + artist_positions[(i + 1) % len(artist_positions)] + i) % 5)
    transaction_records.append({
        "customer": customer,
        "product": product,
        "quantity": quantity,
        "unit_price": product_prices[product],
        "category": "SUPPLIES" if product in ["NOTEBOOK", "PEN"] else "ELECTRONICS"
    })

transaction_records.append(transaction_records[0].copy())

transactions_df = pd.DataFrame(transaction_records)
transactions_df["transaction_value"] = transactions_df["quantity"] * transactions_df["unit_price"]

print("GENERATED TRANSACTION RECORDS")
print(transactions_df.to_string(index=False))

organized_transactions = transactions_df.sort_values(
    ["customer", "product"]
).reset_index(drop=True)
print("\nORGANIZED TRANSACTION DATA")
print(organized_transactions.to_string(index=False))

customer_product_counts = Counter(zip(transactions_df["customer"], transactions_df["product"]))
repeated_purchases = {
    pair: count for pair, count in customer_product_counts.items() if count > 1
}
print("\nREPEATED CUSTOMER/PRODUCT TRANSACTIONS")
print(repeated_purchases)
print("\nDISTINCT CUSTOMERS")
print(sorted(transactions_df["customer"].unique()))
print("\nDISTINCT PRODUCTS")
print(sorted(transactions_df["product"].unique()))

print("\nTRANSACTION VALUES")
print(transactions_df[["customer", "product", "quantity", "unit_price", "transaction_value"]].to_string(index=False))

customer_summary = transactions_df.groupby("customer", as_index=False).agg(
    transaction_count=("transaction_value", "size"),
    total_quantity=("quantity", "sum"),
    total_spent=("transaction_value", "sum")
).sort_values("customer")
product_summary = transactions_df.groupby("product", as_index=False).agg(
    transaction_count=("transaction_value", "size"),
    total_quantity=("quantity", "sum"),
    total_sales=("transaction_value", "sum")
).sort_values("product")

print("\nCUSTOMER SUMMARIES")
print(customer_summary.to_string(index=False, formatters={"total_spent": "{:.2f}".format}))
print("\nPRODUCT SUMMARIES")
print(product_summary.to_string(index=False, formatters={"total_sales": "{:.2f}".format}))

ordered_transactions = transactions_df.sort_values(
    ["customer", "product", "transaction_value"]
).reset_index(drop=True)
print("\nORDERED RESULTS")
print(ordered_transactions.to_string(index=False))

print("\nOVERALL TRANSACTION SUMMARY")
print(f"Number of transaction records: {len(transactions_df)}")
print(f"Distinct customers: {transactions_df['customer'].nunique()}")
print(f"Distinct products: {transactions_df['product'].nunique()}")
print(f"Total quantity sold: {transactions_df['quantity'].sum()}")
print(f"Overall transaction value: {transactions_df['transaction_value'].sum():.2f}")
print(f"Repeated customer/product pairs: {repeated_purchases}")

print("\nFINAL CONSOLIDATED TRANSACTION REPORT")
print("Customer totals:")
print(customer_summary.to_string(index=False, formatters={"total_spent": "{:.2f}".format}))
print("\nProduct totals:")
print(product_summary.to_string(index=False, formatters={"total_sales": "{:.2f}".format}))
print(f"\nGrand total: {transactions_df['transaction_value'].sum():.2f}")
```
